# Radar charts for the current benchmark round

Draws one radar chart per benchmark for the models listed in
[README.md](README.md#-results), writing PNGs into `charts/`.

Datasets covered: CLIcK, HAE-RAE, KoBALT-700, KMMLU-Pro, MuSR(Ko).


```python
import glob
import os
import importlib

from util import radar_chart_generator
importlib.reload(radar_chart_generator)
from util.radar_chart_generator import RadarChartGenerator

# Restrict the comparison to this round's models. Substring match, so the single
# 'deepseek-v4-flash' entry picks up the none/high/max variants.
MODEL_FILTER = [
    "claude-opus-5",
    "claude-sonnet-5",
    "gpt-5.6",
    "deepseek-v4-flash",
]

# Charts stay readable up to roughly this many models per plot.
TOP_N = 10

# Note: the generator loads NanumGothic itself so Korean category labels render.
# Do not override plt.rcParams['font.family'] here — that is what used to turn the
# Korean labels into boxes.
generator = RadarChartGenerator(results_dir="./results")
os.makedirs("./charts", exist_ok=True)
```

## Available result files


```python
for dataset in ["CLIcK", "HAERAE", "KoBALT", "KMMLU-Pro", "MuSR-Ko"]:
    results = generator.read_dataset_results(dataset, model_filter=MODEL_FILTER)
    print(f"{dataset:11s} {len(results)} models")
```

## CLIcK


```python
results = generator.read_dataset_results("CLIcK", model_filter=MODEL_FILTER)
print(f"Found {len(results)} models for CLIcK")

if results:
    generator.create_radar_chart(
        generator.process_click_results(results),
        "CLIcK Performance by Category",
        "./charts/CLIcK_radar_chart.png",
        top_n=TOP_N,
    )
else:
    print("No CLIcK results found")
```
